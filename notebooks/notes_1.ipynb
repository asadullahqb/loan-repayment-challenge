## A. Business Understanding (40% effort)
Underwriting is a keyword that I don't fully understand. Let's study it deeper.
OK I made a naive guess -- what I understand in framing this as a Data Science problem is that underwriting relates strongly to a model that predicts interest rate. The lender might decide to approve or reject the application based the interest rate assigned to an applicant.
However, my guess, was largely wrong. Data Science in underwriting is actually done a different way in the industry, where there is a mixture of rules-based and machine learning models being applied.
- It's wrong to directly jump ahead to interest rate... we should first understand the business problem better, and how established industry players solve the problem.
- I found a book and read a bit of it to learn (i.e. Credit Risk Analytics by Bart Baesens), in it, it is explained and learned for further development purpose that the industry standard way of doing credit risk modelling is to use Machine Learning to predict 3 key metrics i.e. Probability of Default (PD), Loss Given Default (LGD),and Exposure at Default (EAD), where the credit risk score, is a score accounting for these 3 metrics combined. Before an applicant is screened to these credit risk scores, applicants are pre-rejected on the basis of fraud detection, and eligibility checks which save time and resources needed to compile more data.

## D.Model Building (10% effort)
Getting some preliminary ideas on model building where we can start by building a linear regression model to predict interest rate. But this "interest rate" model should be shelved because it is not an established way to approach the problem because it is very naive and oversimplifies the problem.
A better approach is a function of the expected loss:

    ```
    Expected Loss (EL) = PD × LGD × EAD
    ```

I also had an idea to use Large Models and fine-tune them to predict the credit risk score. This approach is more advanced, and requires more data to train. However, it should be avoided because of a few reasons
1. Poor explainability
2. Grinsztajn benchmark still evaluates GBM models as the best model for this problem.

## E. Model Evaluation (10% effort)
We need to evaluate the model based on metrics that make sense. Lack of domain expertise suggests reading for this problem.
Since data for this problem is usually imbalanced, we cannot use accuracy as it just means the model is good at throwing everything as the majority class.
We should instead use ROC-AUC to evaluate the model. Anything less than 70% is rejected as a bad model and we optimise beyond that with a target of 80%. Anything more than that is extra credit.
We should also explore Kolmogorov-Smirnov (KS) statistic to see how well the model separates two groups. It should be able to discern the good and bad borrowers well. We need a 30% KS statistic to be considered a decent model, aim to exceed 40%. 50% or above is a suspicious model.

Further to that, we will then run a Monte Carlo simulation to evaluate the model's profitability. We then plot the full profit distribution and include tail risk.
