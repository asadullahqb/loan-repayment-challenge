# CRISP-DM - Iteration 3
1. Business Understanding (5% effort)
There are established ways to solve this problem and we should not reinvent the wheel or deviate too strongly from them without the strong domain expertise to do so.
The problem should be framed as a credit-scoring pipeline with the following steps:
i. Fraud detection
ii. Eligibility check
iii. Credit scoring (Expected Loss) comprising of...
   a. Probability of Default
   b. Loss Given Default
   c. Exposure at Default


2. Data Understanding (30% effort) -- EDA
Given the problem is framed as a pipeline, we need to identify which data can enable each step of the pipeline.
We should eliminate steps, that cannot be supported by the data available.
i. Fraud detection -- the data presumes this is outsourced to a third-party service. Hence, we do not try to predict fraud ourselves. We may however, design fraud rules based on the data available to eliminate obvious fraud and only train on data where fraud is either not obvious or is absent.
ii. Eligibility check -- the data presumes all applicants are eligible.
iii. Credit scoring (Expected Loss) comprising of...
   a. Probability of Default -- 
   b. Loss Given Default -- the current data is onl 17k defaulted loans, hence we cannot compute this on its own.
   c. Exposure at Default -- since these are fixed-term installment loans, EAD is not applicable. Hence, we will not compute this in the credit scoring and larger pipeline.

For deployment, a Fast API with a Streamlit application should be sufficient for first version.

This should be the animation of the simulation, but not the actual game, the game should still be applicant risk and profitability assessment.
